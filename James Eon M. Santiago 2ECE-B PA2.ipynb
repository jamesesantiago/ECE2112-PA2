```python
import numpy as np
```

(A) Reproducible Normalization Problem

```python
np.random.seed(2112)
X = np.random.randint(10 ,101, size=(5,5))
X
```

```python
X_normalized = (X-X.mean())/X.std()
print("Normalized X\n",X_normalized)
np.save('X_normalized', X_normalized)
```

```python
print("Normalized X Mean:" , X_normalized.mean())
```

```python
print("Normalized X Standard Deviation:",X_normalized.std())
```

(B) Cubes Divisible by 4 Problem

```python
C = np.arange(1,101,1)
C = C**3
C.reshape(10,10)
C
div_by_4= C[C%4==0]

print("Cubes Divisible By 4", div_by_4)
np.save('div_by_4', div_by_4)
```

(C) Above-Mean Squares Problem

```python
S = np.arange(1,37,1)
S = S**2
S_mean = S.mean()
above_mean = S[S>S_mean]
print("Squares Above Mean: \n",above_mean)
np.save('above_mean', above_mean)
```
